```python
RECORDS = (0, 1, 3, 4)
COURSES = ("LC Men", "LC Women" , "SC Men", "SC Women")
list (zip(RECORDS, COURSES))
```

```python
records = {}
for table, course in zip(RECORDS, COURSES):
    records[course] = {}
    for row in tables[table].find("tr", mode="all")[1:]:
        coloums = row.find("td",mode="all")
        event = coloums[0].text
        time = coloums[1].text
        if "relay" not in event:
            records[course][event] = time
      
```

```python
conversions = {
    "Free": "freestyle",
    "Back": "backstroke",
    "Breast": "breaststroke",
    "Fly": "butterfly",
    "IM": "individual medley",
}

event = "Mike-15-100m-Fly.txt"

*_, distance, stroke = event.removesuffix(".txt").split("-")

lookup = f"{distance} {conversions[stroke]}"

print(lookup)
```

```python
for course in records:
    if lookup in records[course]:
        print(records[course][lookup])
    else:
        print(f"{lookup} not found in {course}")
```

```python
def event_lookup (file):
    conversions = {
    "Free": "freestyle",
    "Back": "backstroke",
    "Breast": "breaststroke",
    "Fly": "butterfly",
    "IM": "individual medley",
}
    *_, distance, stroke = file.removesuffix(".txt").split("-")

    return f"{distance} {conversions[stroke]}"

print(event_lookup("Darius-13-100m-IM.txt"))
print(event_lookup(("Darius-13-100m-Fly.txt")))
```

```python
with open ("records.json") as jf:
    records = json.load(jf)
records["LC Men"][event_lookup("Darius-13-100m-Fly.txt")]
```
